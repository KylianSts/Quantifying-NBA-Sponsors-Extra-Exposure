```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
import ipywidgets as widgets
from IPython.display import display

import warnings
warnings.filterwarnings("ignore")

plt.rcParams['figure.figsize'] = (12, 6)
sns.set_theme(style="whitegrid")
pd.set_option('display.max_columns', None)
```

```python
df_exposure = pd.read_csv("Data/exposure_and_game_info/exposure_results_2025_26.csv")
df_metadata = pd.read_csv("Data/urls/game_highlight_urls_2025_26.csv")
df_game_info = pd.read_csv("Data/exposure_and_game_info/nba_games_2025-26.csv")

df = pd.merge(df_exposure, df_metadata.drop(columns=['game_id', 'title', 'url']), on="video_id", how='left')
df = pd.merge(df, df_game_info[['GAME_ID', 'HOME_TEAM_ID', "HOME_TEAM_NAME", 'AWAY_TEAM_ID', "AWAY_TEAM_NAME", 'GAME_DATE']], left_on='game_id', right_on='GAME_ID', how='left')

display(df.head())
print(df.shape)
```

```python
# =============================================================================
# DATA PREPARATION
# =============================================================================

# Calculate number of URLs per game
nb_urls_per_game = df.groupby('game_id')['video_id'].count()
df['nb_urls_per_game'] = df['game_id'].map(nb_urls_per_game)

# Convert dates
df['GAME_DATE'] = pd.to_datetime(df['GAME_DATE'])

# Basic calculations
nb_games = df['game_id'].nunique()
nb_days = (df['GAME_DATE'].max() - df['GAME_DATE'].min()).days
games_per_day = nb_games / nb_days

# DataFrame with unique videos (to avoid duplicates in calculations)
df_videos = df.drop_duplicates(subset='video_id')


# =============================================================================
# GENERAL STATISTICS
# =============================================================================

print("=" * 50)
print("          GENERAL STATISTICS")
print("=" * 50)

print("\nCOVERAGE")
print(f"  Number of games covered: {nb_games:,}")
print(f"  Number of videos analyzed: {df['video_id'].nunique():,}")

print("\nTOTAL ENGAGEMENT")
print(f"  Total views      : {df_videos['view_count'].sum():>15,.0f}")
print(f"  Total likes      : {df_videos['like_count'].sum():>15,.0f}")
print(f"  Total comments   : {df_videos['comment_count'].sum():>15,.0f}")

print("\nAVERAGES PER GAME")
print(f"  Views/game       : {df_videos['view_count'].sum() / nb_games:>15,.0f}")
print(f"  Likes/game       : {df_videos['like_count'].sum() / nb_games:>15,.0f}")
print(f"  Comments/game    : {df_videos['comment_count'].sum() / nb_games:>15,.0f}")

print("\nDAILY AVERAGES")
print(f"  Games/day        : {games_per_day:>15.2f}")
views_per_day = (df_videos['view_count'].sum() / nb_games) * games_per_day
likes_per_day = (df_videos['like_count'].sum() / nb_games) * games_per_day
comments_per_day = (df_videos['comment_count'].sum() / nb_games) * games_per_day
print(f"  Views/day        : {views_per_day:>15,.0f}")
print(f"  Likes/day        : {likes_per_day:>15,.0f}")
print(f"  Comments/day     : {comments_per_day:>15,.0f}")
print("=" * 50)


# =============================================================================
# VISUALIZATIONS
# =============================================================================

# 1. Distribution of views (Logarithmic scale)
plt.figure(figsize=(12, 6))
sns.histplot(df_videos['view_count'], bins=50, log_scale=True, color='teal')
plt.title("Distribution of Views per Video (Log Scale)", fontsize=14, fontweight='bold')
plt.xlabel("Number of Views")
plt.ylabel("Number of Videos")
plt.tight_layout()
plt.show()

# 2. Number of videos per game over time
plt.figure(figsize=(12, 6))
sns.lineplot(x='GAME_DATE', y='nb_urls_per_game', data=df, errorbar=None, color='steelblue')
plt.title("Number of Highlights per Game Over Time", fontsize=14, fontweight='bold')
plt.xlabel("Game Date")
plt.ylabel("Number of Videos")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 3. Temporal evolution of view counts
plt.figure(figsize=(12, 6))
sns.lineplot(x=df_videos['GAME_DATE'], y=df_videos['view_count'], color='teal')
plt.title("Evolution of Views per Video", fontsize=14, fontweight='bold')
plt.xlabel("Game Date")
plt.ylabel("Number of Views")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# =============================================================================
# TOP GAMES
# =============================================================================

print("\n" + "=" * 50)
print("          TOP 10 MOST VIEWED GAMES")
print("=" * 50)

top_games = (df_videos.groupby(['HOME_TEAM_NAME', 'AWAY_TEAM_NAME'])
             .agg({
                 'view_count': 'sum', 
                 'like_count': 'sum',
                 'comment_count': 'sum',
                 'GAME_DATE': 'first'
             })
             .sort_values(by='view_count', ascending=False)
             .head(10))

display(top_games)
```

```python
# =============================================================================
# EXPOSURE TIME ANALYSIS BY ZONE
# =============================================================================

# Calculate exposure percentage
df['percentage_exposition'] = (df['exposure_seconds'] / df['duration']) * 100

# Sort order for charts (from most exposed to least exposed)
order_by_time = (df.groupby("exposure_zone")["exposure_seconds"]
                 .median()
                 .sort_values(ascending=False)
                 .index)


# =============================================================================
# SIDE-BY-SIDE VISUALIZATIONS
# =============================================================================

# Create side-by-side subplots
fig, axes = plt.subplots(1, 2, figsize=(20, 6))

# Chart 1: Absolute exposure time (in seconds)
sns.boxplot(x="exposure_zone", y="exposure_seconds", data=df, 
            order=order_by_time, palette="viridis", ax=axes[0])
axes[0].set_title("Absolute Exposure Time", fontsize=14, fontweight='bold')
axes[0].set_xlabel("Exposure Zone")
axes[0].set_ylabel("Exposure Time (seconds)")
axes[0].tick_params(axis='x', rotation=45, labelrotation=45)
for label in axes[0].get_xticklabels():
    label.set_ha('right')

# Chart 2: Relative exposure time (in percentage)
sns.boxplot(x="exposure_zone", y="percentage_exposition", data=df, 
            order=order_by_time, palette="viridis", ax=axes[1])
axes[1].set_title("Relative Exposure Time", fontsize=14, fontweight='bold')
axes[1].set_xlabel("Exposure Zone")
axes[1].set_ylabel("Percentage of Video (%)")
axes[1].tick_params(axis='x', rotation=45, labelrotation=45)
for label in axes[1].get_xticklabels():
    label.set_ha('right')

# Overall title
plt.suptitle("Distribution of Exposure Time by Zone", 
             fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()


# =============================================================================
# OUTLIER ANALYSIS
# =============================================================================

print("\n" + "=" * 70)
print("     INVESTIGATION: Videos with Low Basketball-Logo Exposure")
print("=" * 70)

# Analysis of videos with very low basketball logo exposure
low_exposure = (df[df['exposure_zone'] == 'basket-logo']
                [['percentage_exposition', 'video_id', 'view_count']]
                .sort_values(by='percentage_exposition')
                .head(10))

display(low_exposure)
print("=" * 70)
```

```python
# =============================================================================
# COMPARISON: WEIGHTED VS UNWEIGHTED EXPOSURE
# =============================================================================

# 1. Calculate UNWEIGHTED average (percentage)
df_unweighted_pct = df.groupby(['HOME_TEAM_NAME', 'exposure_zone'])['percentage_exposition'].mean()
df_heatmap_unweighted_pct = df_unweighted_pct.reset_index().pivot(
    index="HOME_TEAM_NAME", 
    columns="exposure_zone", 
    values="percentage_exposition"
)

# 2. Calculate WEIGHTED average by views (percentage)
df_weighted_pct = (df.groupby(['HOME_TEAM_NAME', 'exposure_zone'])
                   .apply(lambda x: (x['percentage_exposition'] * x['view_count']).sum() / x['view_count'].sum())
                   .reset_index(name='weighted_percentage'))

df_heatmap_weighted_pct = df_weighted_pct.pivot(
    index="HOME_TEAM_NAME", 
    columns="exposure_zone", 
    values="weighted_percentage"
)

# 3. Calculate UNWEIGHTED average (duration)
df_unweighted_dur = df.groupby(['HOME_TEAM_NAME', 'exposure_zone'])['duration'].mean()
df_heatmap_unweighted_dur = df_unweighted_dur.reset_index().pivot(
    index="HOME_TEAM_NAME", 
    columns="exposure_zone", 
    values="duration"
)

# 4. Calculate WEIGHTED average by views (duration)
df_weighted_dur = (df.groupby(['HOME_TEAM_NAME', 'exposure_zone'])
                   .apply(lambda x: (x['duration'] * x['view_count']).sum() / x['view_count'].sum())
                   .reset_index(name='weighted_duration'))

df_heatmap_weighted_dur = df_weighted_dur.pivot(
    index="HOME_TEAM_NAME", 
    columns="exposure_zone", 
    values="weighted_duration"
)

# 5. 2x2 Visualization
fig, axes = plt.subplots(2, 2, figsize=(24, 20))

# Top Left: Unweighted Percentage
sns.heatmap(
    df_heatmap_unweighted_pct,
    cmap="YlGnBu",
    annot=True,
    fmt=".1f",
    linewidths=0.5,
    cbar_kws={"label": "% of video time"},
    ax=axes[0, 0]
)
axes[0, 0].set_title("Simple Average - Percentage (Unweighted)", fontsize=14, fontweight='bold', pad=20)
axes[0, 0].set_ylabel("Home Team")
axes[0, 0].set_xlabel("Exposure Zone")

# Top Right: Weighted Percentage
sns.heatmap(
    df_heatmap_weighted_pct,
    cmap="YlGnBu",
    annot=True,
    fmt=".1f",
    linewidths=0.5,
    cbar_kws={"label": "% of video time (weighted)"},
    ax=axes[0, 1]
)
axes[0, 1].set_title("View-Weighted Average - Percentage", fontsize=14, fontweight='bold', pad=20)
axes[0, 1].set_ylabel("Home Team")
axes[0, 1].set_xlabel("Exposure Zone")

# Bottom Left: Unweighted Duration
sns.heatmap(
    df_heatmap_unweighted_dur,
    cmap="YlGnBu",
    annot=True,
    fmt=".1f",
    linewidths=0.5,
    cbar_kws={"label": "Duration (seconds)"},
    ax=axes[1, 0]
)
axes[1, 0].set_title("Simple Average - Duration (Unweighted)", fontsize=14, fontweight='bold', pad=20)
axes[1, 0].set_ylabel("Home Team")
axes[1, 0].set_xlabel("Exposure Zone")

# Bottom Right: Weighted Duration
sns.heatmap(
    df_heatmap_weighted_dur,
    cmap="YlGnBu",
    annot=True,
    fmt=".1f",
    linewidths=0.5,
    cbar_kws={"label": "Duration (seconds, weighted)"},
    ax=axes[1, 1]
)
axes[1, 1].set_title("View-Weighted Average - Duration", fontsize=14, fontweight='bold', pad=20)
axes[1, 1].set_ylabel("Home Team")
axes[1, 1].set_xlabel("Exposure Zone")

plt.suptitle("Exposure Intensity by Zone According to Home Team", 
             fontsize=16, fontweight='bold', y=1.00)
plt.tight_layout()
plt.show()
```

```python
# =============================================================================
# SIZE MODEL (SIGMOID)
# =============================================================================

def plot_size_model(k=120, m=0.008):
    """
    Displays sigmoid score curve as a function of logo relative size.
    k : Steepness - Higher values create sharper transitions
    m : Midpoint - The size at which score is 0.5
    """
    
    # Create data (Ratio from 0% to 5% of screen)
    # Focus on small values where the action happens
    x_values = np.linspace(0, 0.05, 500) 
    
    # Sigmoid Formula: S(x) = 1 / (1 + exp(-k * (x - m)))
    s_values = 1 / (1 + np.exp(-k * (x_values - m)))
    
    # Graph configuration
    plt.figure(figsize=(10, 6))
    
    # Plot the curve
    plt.plot(x_values, s_values, lw=3, color='#3498db', label=f'Score (k={k}, m={m:.3f})')
    
    # Reference zones (Based on NBA data)
    # Basket Logo (~0.2%)
    plt.axvline(x=0.002, color='red', linestyle='--', alpha=0.5)
    plt.text(0.0022, 0.1, 'Basket (0.2%)', color='red', rotation=90)
    
    # Side Court (~0.8%)
    plt.axvline(x=0.008, color='orange', linestyle='--', alpha=0.5)
    plt.text(0.0082, 0.1, 'Side (0.8%)', color='orange', rotation=90)
    
    # Back Court (~1.5%)
    plt.axvline(x=0.015, color='green', linestyle='--', alpha=0.5)
    plt.text(0.0152, 0.1, 'Back (1.5%)', color='green', rotation=90)

    # Mid Court (~2.0%)
    plt.axvline(x=0.018, color='purple', linestyle='--', alpha=0.5)
    plt.text(0.0202, 0.1, 'Mid (1.8%)', color='purple', rotation=90)

    # Key points on the curve for references
    def get_score(x, k, m): 
        return 1 / (1 + np.exp(-k * (x - m)))
    
    score_basket = get_score(0.002, k, m)
    plt.scatter([0.002], [score_basket], color='red', zorder=5)
    plt.text(0.0025, score_basket, f'{score_basket:.2f}', color='red', fontweight='bold')

    score_back = get_score(0.015, k, m)
    plt.scatter([0.015], [score_back], color='green', zorder=5)
    plt.text(0.0155, score_back, f'{score_back:.2f}', color='green', fontweight='bold')

    # Styling
    plt.title("Size Score Modeling (Sigmoid)", fontsize=14)
    plt.xlabel("Logo Surface (% of screen)", fontsize=12)
    plt.ylabel("Size Score (0 to 1)", fontsize=12)
    plt.xlim(0, 0.035)  # Display up to 3.5%
    plt.ylim(0, 1.1)
    plt.grid(True, linestyle='--', alpha=0.6)
    
    # Format X-axis as percentage
    plt.gca().xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: '{:.1%}'.format(x)))
    
    plt.legend(loc='lower right')
    plt.show()

# 2. Create Interactive Sliders
style = {'description_width': 'initial'}

slider_k_size = widgets.IntSlider(
    value=120,
    min=20,
    max=500,
    step=10,
    description='Steepness (k):',
    style=style,
    layout=widgets.Layout(width='600px')
)

slider_m_size = widgets.FloatSlider(
    value=0.008,
    min=0.0005,
    max=0.03,
    step=0.0005,
    description='Midpoint (m):',
    readout_format='.4f',
    style=style,
    layout=widgets.Layout(width='600px')
)

# 3. Display
ui = widgets.VBox([slider_k_size, slider_m_size])
out = widgets.interactive_output(plot_size_model, {'k': slider_k_size, 'm': slider_m_size})

display(ui, out)
```

```python
# =============================================================================
# ATTENTION MODEL (GAUSSIAN)
# =============================================================================

def plot_attention_model(sigma=0.20):
    """
    Displays attention score as a function of distance from focus point.
    sigma : Width of attention focus (standard deviation)
    """
    
    # Create data
    d_values = np.linspace(0, 1.0, 500)  # Distance from 0% to 100% of diagonal
    
    # Gaussian Formula (Attention Score)
    # Avoid division by zero
    s = max(0.01, sigma)
    s_values = np.exp(- (d_values**2) / (2 * s**2))
    
    # Graph configuration
    plt.figure(figsize=(10, 6))
    
    # Reference zones (Basketball context)
    plt.axvspan(0, 0.05, color='green', alpha=0.15, label='Fovea (Pure Focus - Ball)')
    plt.axvspan(0.05, 0.20, color='yellow', alpha=0.15, label='Parafovea (Near Play)')
    plt.axvspan(0.20, 1.0, color='gray', alpha=0.1, label='Periphery (Blurred Vision)')
    
    # Plot the curve
    plt.plot(d_values, s_values, lw=3, color='#FF5733', label=f'Score (Sigma={sigma:.2f})')
    
    # Key points
    # Example: Score at 20% distance
    score_at_20 = np.exp(- (0.20**2) / (2 * s**2))
    plt.scatter([0.20], [score_at_20], color='black', zorder=5)
    plt.text(0.21, score_at_20 + 0.02, f'Score @20%: {score_at_20:.2f}', fontsize=10)

    # Styling
    plt.title(f"Scientific Attention Modeling (Sigma = {sigma:.2f})", fontsize=14)
    plt.xlabel("Distance from Ball (% of screen diagonal)", fontsize=12)
    plt.ylabel("Attention Score (0 to 1)", fontsize=12)
    plt.xlim(0, 1.0)
    plt.ylim(0, 1.1)
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend(loc='upper right')
    plt.show()

# 2. Create Interactive Slider
style = {'description_width': 'initial'}

slider_sigma_attention = widgets.FloatSlider(
    value=0.20,
    min=0.05,
    max=0.50,
    step=0.01,
    description='Sigma (Focus Width):',
    style=style,
    layout=widgets.Layout(width='600px')
)

# 3. Display
ui = widgets.VBox([slider_sigma_attention])
out = widgets.interactive_output(plot_attention_model, {'sigma': slider_sigma_attention})

display(ui, out)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# =============================================================================
# MULTI-COMPETITOR SOV CLUTTER MODEL - IMPROVED UI
# =============================================================================

def plot_sov_multi_model(sigma=0.10, use_sqrt=True, num_competitors=1,
                         size1=1.0, dist1=0.15, size2=1.0, dist2=0.25, 
                         size3=1.0, dist3=0.35, size4=1.0, dist4=0.40,
                         size5=1.0, dist5=0.45):
    """
    Simulates the Clutter score of a logo facing multiple competitors.
    """
    
    # Gather active competitors
    competitors = []
    if num_competitors >= 1:
        competitors.append({'size': size1, 'dist': dist1, 'color': '#e74c3c', 'name': 'Comp 1'})
    if num_competitors >= 2:
        competitors.append({'size': size2, 'dist': dist2, 'color': '#3498db', 'name': 'Comp 2'})
    if num_competitors >= 3:
        competitors.append({'size': size3, 'dist': dist3, 'color': '#f39c12', 'name': 'Comp 3'})
    if num_competitors >= 4:
        competitors.append({'size': size4, 'dist': dist4, 'color': '#9b59b6', 'name': 'Comp 4'})
    if num_competitors >= 5:
        competitors.append({'size': size5, 'dist': dist5, 'color': '#1abc9c', 'name': 'Comp 5'})
    
    # Create distance range
    x_values = np.linspace(0, 0.5, 500)
    
    # Calculate total weighted competitor area
    s = max(0.01, sigma)
    total_weighted_area = np.zeros_like(x_values)
    
    for comp in competitors:
        weight = np.exp(- ((x_values - comp['dist'])**2) / (2 * s**2))
        total_weighted_area += comp['size'] * weight
    
    # Calculate Share of Voice
    sov_raw = 1.0 / (1.0 + total_weighted_area)
    
    # Apply Square Root (Optional)
    if use_sqrt:
        final_score = np.sqrt(sov_raw)
        label_curve = f'My Score (With √, {num_competitors} Competitors)'
        color_curve = '#8e44ad'
    else:
        final_score = sov_raw
        label_curve = f'My Score (Raw, {num_competitors} Competitors)'
        color_curve = '#e74c3c'
    
    # Create figure
    plt.figure(figsize=(14, 8))
    
    # Plot main score curve
    plt.plot(x_values, final_score, lw=4, color=color_curve, label=label_curve, zorder=10)
    
    # Plot raw SoV in dashed line if square root is active
    if use_sqrt:
        plt.plot(x_values, sov_raw, lw=2, color='gray', linestyle='--', 
                label='Raw SoV (Mathematical)', alpha=0.5, zorder=5)
    
    # Plot competitor positions
    for comp in competitors:
        idx = np.abs(x_values - comp['dist']).argmin()
        score_at_comp = final_score[idx]
        plt.axvline(comp['dist'], color=comp['color'], linestyle=':', alpha=0.7, lw=2.5)
        plt.scatter([comp['dist']], [score_at_comp], color=comp['color'], s=200, 
                   zorder=15, edgecolors='black', linewidths=2.5, marker='o')
        plt.text(comp['dist'], score_at_comp - 0.09, 
                f"{comp['name']}\nSize: {comp['size']:.1f}x\nScore: {score_at_comp:.2f}", 
                ha='center', fontsize=10, color='white', fontweight='bold',
                bbox=dict(boxstyle='round,pad=0.5', facecolor=comp['color'], 
                         alpha=0.9, edgecolor='black', linewidth=2))

    # Reference zones with better visibility
    plt.axvspan(0, 0.05, color='red', alpha=0.1, label='Direct Contact (0-0.05)', zorder=1)
    plt.axvspan(0.05, 0.20, color='orange', alpha=0.1, label='Near Zone (0.05-0.20)', zorder=1)
    plt.axvspan(0.20, 0.50, color='green', alpha=0.1, label='Far Zone (0.20-0.50)', zorder=1)

    # Dynamic title
    status = "ON" if use_sqrt else "OFF"
    avg_score = np.mean(final_score)
    plt.title(f"Multi-Competitor Clutter Model ({num_competitors} Competitors)\n" + 
             f"Square Root: {status} | Sigma: {sigma:.2f} | Average Score: {avg_score:.2f}", 
             fontsize=16, fontweight='bold', pad=20)
    plt.xlabel("Distance from me (% of diagonal)", fontsize=14, fontweight='bold')
    plt.ylabel("My Clutter Score (0 to 1)", fontsize=14, fontweight='bold')
    plt.xlim(0, 0.5)
    plt.ylim(0, 1.1)
    plt.grid(True, linestyle='--', alpha=0.4, linewidth=0.8)
    plt.legend(loc='upper right', fontsize=10, framealpha=0.95)
    plt.tight_layout()
    plt.show()


# =============================================================================
# IMPROVED WIDGET LAYOUT
# =============================================================================

# Improved styling
header_style = {'description_width': '200px'}
slider_layout = widgets.Layout(width='500px')

# Custom CSS for better readability
display(widgets.HTML("""
<style>
    .widget-label { font-size: 14px !important; font-weight: 600 !important; }
    .widget-slider { margin: 10px 0 !important; }
    .jupyter-widgets { font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', sans-serif !important; }
</style>
"""))

# ============= GLOBAL PARAMETERS =============


slider_sigma = widgets.FloatSlider(
    value=0.10, min=0.01, max=0.50, step=0.01,
    description='Sigma (Influence Radius):',
    style=header_style, layout=slider_layout,
    readout_format='.2f'
)

slider_num = widgets.IntSlider(
    value=1, min=1, max=5, step=1,
    description='Number of Competitors:',
    style=header_style, layout=slider_layout
)

checkbox_sqrt = widgets.Checkbox(
    value=True,
    description='Enable Square Root Smoothing (Stevens\' Law)',
    style={'description_width': 'initial'},
    indent=False
)

global_box = widgets.VBox([
    widgets.HTML("<h3 style='color: #2c3e50; margin: 10px 0;'>⚙️ Global Settings</h3>"),
    slider_sigma,
    widgets.HTML("<p style='color: #7f8c8d; margin: 5px 0 15px 0; font-size: 12px;'>↑ Controls how far competitor influence extends</p>"),
    slider_num,
    checkbox_sqrt
], layout=widgets.Layout(padding='15px', border='2px solid #3498db', border_radius='10px', margin='10px'))

# ============= COMPETITOR CONTROLS =============

def create_competitor_controls(num, color, default_size=1.0, default_dist=0.15):
    """Create a styled competitor control group"""
    
    size_slider = widgets.FloatSlider(
        value=default_size, min=0.1, max=10.0, step=0.1,
        description=f'Size (relative):',
        style=header_style, layout=slider_layout,
        readout_format='.1f'
    )
    
    dist_slider = widgets.FloatSlider(
        value=default_dist, min=0.0, max=0.5, step=0.01,
        description=f'Distance (% diagonal):',
        style=header_style, layout=slider_layout,
        readout_format='.2f'
    )
    
    return widgets.VBox([
        widgets.HTML(f"<h4 style='color: {color}; margin: 5px 0; padding: 8px; background-color: {color}15; border-left: 4px solid {color}; border-radius: 5px;'>🔸 Competitor {num}</h4>"),
        size_slider,
        dist_slider
    ], layout=widgets.Layout(margin='10px 0'))

# Create all competitor controls
comp1_controls = create_competitor_controls(1, '#e74c3c', 1.0, 0.15)
comp2_controls = create_competitor_controls(2, '#3498db', 1.0, 0.25)
comp3_controls = create_competitor_controls(3, '#f39c12', 1.0, 0.35)
comp4_controls = create_competitor_controls(4, '#9b59b6', 1.0, 0.40)
comp5_controls = create_competitor_controls(5, '#1abc9c', 1.0, 0.45)

# Extract individual sliders for the interactive output
size1, dist1 = comp1_controls.children[1], comp1_controls.children[2]
size2, dist2 = comp2_controls.children[1], comp2_controls.children[2]
size3, dist3 = comp3_controls.children[1], comp3_controls.children[2]
size4, dist4 = comp4_controls.children[1], comp4_controls.children[2]
size5, dist5 = comp5_controls.children[1], comp5_controls.children[2]

competitors_box = widgets.VBox([
    widgets.HTML("<h3 style='color: #2c3e50; margin: 10px 0;'>🎯 Competitor Configuration</h3>"),
    comp1_controls,
    comp2_controls,
    comp3_controls,
    comp4_controls,
    comp5_controls
], layout=widgets.Layout(padding='15px', border='2px solid #e74c3c', border_radius='10px', margin='10px'))

# ============= LAYOUT =============
# Create tabs for better organization
tab = widgets.Tab()
tab.children = [global_box, competitors_box]
tab.set_title(0, '⚙️ Global Settings')
tab.set_title(1, '👥 Competitors')

# Add info box
info_box = widgets.HTML("""
<div style='background-color: #ecf0f1; padding: 15px; border-radius: 10px; margin: 10px 0; border-left: 5px solid #3498db;'>
    <h4 style='margin-top: 0; color: #2c3e50;'>ℹ️ How to Use:</h4>
    <ul style='color: #34495e; line-height: 1.8;'>
        <li><strong>Sigma:</strong> Adjust how much distance affects competitor influence</li>
        <li><strong>Number of Competitors:</strong> Select how many competitors to analyze (1-5)</li>
        <li><strong>Size:</strong> Set each competitor's relative size compared to yours</li>
        <li><strong>Distance:</strong> Position competitors at different distances from you</li>
        <li><strong>Square Root:</strong> Apply perceptual smoothing based on Stevens' Law</li>
    </ul>
</div>
""")

# Display everything
display(info_box)
display(tab)

# Create interactive output
out = widgets.interactive_output(
    plot_sov_multi_model,
    {
        'sigma': slider_sigma,
        'use_sqrt': checkbox_sqrt,
        'num_competitors': slider_num,
        'size1': size1, 'dist1': dist1,
        'size2': size2, 'dist2': dist2,
        'size3': size3, 'dist3': dist3,
        'size4': size4, 'dist4': dist4,
        'size5': size5, 'dist5': dist5
    }
)

display(out)
```

```python
# 1. Sigmoid Function (Readability/Laplace Model)
def plot_readability_model(k=0.05, m=100):
    """
    Displays the score curve as a function of Laplace Variance (Sharpness).
    k : Steepness - Must be small because X scale is large (0-500)
    m : Midpoint - The variance required to achieve 0.5 (Average)
    """
    
    # Create data (Variance from 0 to 500)
    # 0 = Solid gray (No edges)
    # 500 = Very sharp contours (4K, White text on black)
    x_values = np.linspace(0, 500, 500) 
    
    # Sigmoid Formula: S(x) = 1 / (1 + exp(-k * (x - m)))
    # Note: Negative sign because higher variance is better
    s_values = 1 / (1 + np.exp(-k * (x_values - m)))
    
    # Graph configuration
    plt.figure(figsize=(10, 6))
    
    # Plot the curve
    plt.plot(x_values, s_values, lw=3, color='#2ecc71', label=f'Score (k={k:.3f}, m={m})')
    
    # Reference zones (Computer Vision Context)
    plt.axvspan(0, 50, color='red', alpha=0.1, label='Blurry / Unreadable')
    plt.axvspan(50, 150, color='orange', alpha=0.1, label='Acceptable')
    plt.axvspan(150, 500, color='green', alpha=0.1, label='High Definition')

    # Key points on the curve
    def get_score(x, k, m): 
        return 1 / (1 + np.exp(-k * (x - m)))
    
    # Test at 50 (Lower limit)
    s_50 = get_score(50, k, m)
    plt.scatter([50], [s_50], color='red', zorder=5)
    plt.text(55, s_50, f'Var 50: {s_50:.2f}', color='red', fontsize=9)

    # Test at 100 (Average target)
    s_100 = get_score(100, k, m)
    plt.scatter([100], [s_100], color='orange', zorder=5)
    plt.text(105, s_100 - 0.05, f'Var 100: {s_100:.2f}', color='orange', fontsize=9, fontweight='bold')

    # Test at 300 (Perfect)
    s_300 = get_score(300, k, m)
    plt.scatter([300], [s_300], color='green', zorder=5)
    plt.text(305, s_300, f'Var 300: {s_300:.2f}', color='green', fontsize=9)

    # Styling
    plt.title("Readability Score Modeling (Laplace Variance)", fontsize=14)
    plt.xlabel("Laplace Variance (Sharpness Indicator)", fontsize=12)
    plt.ylabel("Readability Score (0 to 1)", fontsize=12)
    plt.xlim(0, 400)
    plt.ylim(0, 1.1)
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend(loc='lower right')
    plt.show()

# 2. Create Interactive Sliders
style = {'description_width': 'initial'}

slider_k = widgets.FloatSlider(
    value=0.05,
    min=0.005,
    max=0.30,
    step=0.005,
    description='Steepness (k):',
    style=style,
    layout=widgets.Layout(width='600px')
)

slider_m = widgets.IntSlider(
    value=100,
    min=20,
    max=400,
    step=5,
    description='Midpoint (m):',
    style=style,
    layout=widgets.Layout(width='600px')
)

# 3. Display
ui = widgets.VBox([slider_k, slider_m])
out = widgets.interactive_output(plot_readability_model, {'k': slider_k, 'm': slider_m})

display(ui, out)
```
